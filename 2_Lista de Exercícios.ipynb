
**10. O salario lıquido de um trabalhador e o valor do seu salario bruto, do qual sao descontados
11% para o Imposto de Renda, 8% para o INSS e 5% para o sindicato, faca um programa
em Python que leia o salario bruto e forneca uma saıda como mostrada abaixo:**
```
+ Salario Bruto: R$
- IR (11%): R$
- INSS (8%): R$
- Sindicato (5%): R$
= Salario Lıquido: R

````python
while True:

  #ENTRADA DE DADOS
  salario = input(str("Digite seu salário fixo: "))

  #CONDIÇÕES DE DADOS
  if any(char.isalpha() for char in salario): #CONFERENCIA SE FOR ESCRITO APENAS NÚMEROS
     print("Digite apenas números")
     continue
  else:
    break
if salario == salario :
  salario_total = float(salario)
  ir = salario_total*0.11
  inss = salario_total*0.08
  sindicato = salario_total*0.05
  salario_liquido = salario_total - (ir + inss + sindicato)
  print(f"Salário bruto : R${salario_total:.2f}")
  print(f"IR (11%) : R${ir:.2f}")
  print(f"INSS (8%) : R${inss:.2f}")
  print(f"Sindicato (5%) : R${sindicato:.2f}")
  print(f"Salário líquido : R${salario_liquido:.2f}")
else:
  print("ERRO INESPERADO")
````

**11. Faca um programa em Python para uma loja de tintas. O programa devera pedir o tamanho
em metros quadrados da area a ser pintada. Considere que a cobertura da tinta e de 1 litro
para cada 3 metros quadrados e que a tinta e vendida em latas de 18 litros, que custam
R$ 80,00. Informe ao usuario a quantidades de latas de tinta a serem compradas e o preco
total.**

```
PENSSAMENTO FINAL NINGUEM COMPRA MENOS DE UMA LATA ENTAO PENSSO QUE:
 1 LITRO PINTA 3 METROS
 18 LITROS PINTA 54 METROS

````python
#Repetição até enviar certo
while True:

  #ENTRADA DE DADOS
  area = input(str("Digite a área a ser pintada: "))

  #CONDIÇÕES DE DADOS
  if any(char.isalpha() for char in area): #CONFERENCIA SE FOR ESCRITO APENAS NÚMEROS
     print("Digite apenas numeros")
     continue
  else:
    break
area1 = float(area)
if area1 >= 54:
  cobertura_de_tinta = area1 * 1/3
  quantidades_de_latas = cobertura_de_tinta / 18
  preco_total = quantidades_de_latas * 80
  if quantidades_de_latas <= 1:
    print("Quantidade de latas: 1")
    print("Preço total: R$80,00")
  elif quantidades_de_latas > 1:
   print(f"Quantidade de latas: {quantidades_de_latas:.0f}")
   print(f"Preço total: R${preco_total:.2f}")
elif area1 < 54 and area1 > 0:
   print("Quantidade de latas: 1")
   print("Preço total: R$80,00")
else:
  print("ERRO INESPERADO")


````

**12. Faca um programa em Pyton que peca o tamanho de um arquivo para download (em MB)
e a velocidade de um link de Internet (em Mbps), calcule e informe o tempo aproximado
de download do arquivo usando este link (em minutos). Lembre-se que 1 Byte = 8 bits,
1kB = 1024 Bytes e 1MB = 1024 kB.**

FALTA EU TRANFORMAR DE MINUTOS EM SEGUNDOS

````python
#Continuar pedindo até dar certo
while True:

   #Entrada de dados
   tamanho_do_arquivo = input(str("Digite o tamanho do arquivo em MB: "))
   velocidade_de_internet = input(str("Digite a velocidade da internet em Mbps: "))

   #CONDIÇÕES DE DADOS
   if any(char.isalpha() for char in tamanho_do_arquivo): #CONFERENCIA SE FOR ESCRITO APENAS NÚMEROS
     print("Digite apenas números")
     continue
   else:
     break
tamanho_do_arquivo1 = float(
    tamanho_do_arquivo
    )
velocidade_de_internet1 = float(
    velocidade_de_internet
    )
tempo_de_download = (
    tamanho_do_arquivo1 * 8) / (velocidade_de_internet1)
print(f"O tempo de dowload do arquivo foi: {tempo_de_download:.0f}s")
````
