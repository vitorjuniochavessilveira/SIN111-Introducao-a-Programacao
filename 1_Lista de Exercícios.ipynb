```python
 #1) TRANFORMAÇÃO DE HEXADECIMAL PARA DECIMAl E DEPOIS PARA BINARIO(BYTE)
a = 9
f = 15
#HEXADECIMAL -> DECIMAL
c = a*16**1
d = f*16**0
e= c+d
#DECIAMAL -> BINARIO
try:
  binario = bin(e).replace("0b","")
  print(f"A tranformação de 9F para decimal é {c}+{d}={e}")
  print(f"Agora para binario(byte)= {binario}")
except ValueError:
  print("Hexadecimal Invalido")
```

```python
#3) Converção de Numero Binario para Deacimal e Hexadecimal
# Decimal
a = "1011100"
try:
  inteiro =int(a, 2) #Sinaliza a base do numero 2
  print(f"{a} em decimal inteiro é {inteiro}")
except ValueError:
  print("Binario Invalido")
# Hexadecimal (EU AINDA NAO ESTOU SABENDO FAZER)
b = "1011100"
try:
  hexadecimal = hex(1011100).replace("0x","") #Sinaliza a base do numero 16
  print(f"{b} em hexadecimal é {hexadecimal}")
except ValueError:
        print("Hexadecimal Invalido")

```

```python
#4.2- Proximo Binario
def proximo_binario(a):
  try:
    decimal= int(a,2) #Converter binario para decimal
    decimal += 1 #Incrementa
    return format(decimal, 'b')#Converte de volta para binário
  except ValueError:
    return "Entrada inválida. Use apenas 0 e 1"
print(proximo_binario("10101111"))

```
